```python
import time

def get_matrix(label):
    print(f"\nEnter the {label} state row by row.")
    print("Provide 3 numbers per row separated by spaces. Use '0' for the blank space.")
    matrix = []
    for i in range(1, 4):
        while True:
            try:
                row = tuple(map(int, input(f"Row {i}: ").strip().split()))
                if len(row) == 3:
                    matrix.append(row)
                    break
                else:
                    print("Please enter exactly 3 numbers separated by spaces.")
            except ValueError:
                print("Invalid input. Please enter integers only.")
    return tuple(matrix)

def print_state(state):
    for row in state:
        print(" " + " ".join(map(str, row)))
    print()

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return -1, -1

def get_neighbors(state):
    r, c = find_blank(state)
    # Move order: Up, Down, Left, Right
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    neighbors = []
    for dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            grid = [list(row) for row in state]
            grid[r][c], grid[nr][nc] = grid[nr][nc], grid[r][c]
            neighbors.append(tuple(tuple(row) for row in grid))
    return neighbors

def depth_limited_search(current, goal, limit, path, visited_in_path, stats):
    stats["nodes_explored"] += 1
    stats["max_stack_depth"] = max(stats["max_stack_depth"], len(path))

    if current == goal:
        return path

    if limit <= 0:
        return None

    for neighbor in get_neighbors(current):
        if neighbor not in visited_in_path:
            visited_in_path.add(neighbor)
            result = depth_limited_search(
                neighbor,
                goal,
                limit - 1,
                path + [neighbor],
                visited_in_path,
                stats
            )
            if result is not None:
                return result
            visited_in_path.remove(neighbor)

    return None

def solve_iddfs(initial_state, goal_state, max_depth_limit):
    start_time = time.time()
    stats = {"nodes_explored": 0, "max_stack_depth": 0}
    depth = 0

    while True:
        if max_depth_limit > 0 and depth > max_depth_limit:
            break

        print(f"Searching at depth limit: {depth}...")
        visited_in_path = {initial_state}
        result = depth_limited_search(
            initial_state,
            goal_state,
            depth,
            [initial_state],
            visited_in_path,
            stats
        )

        if result is not None:
            elapsed = time.time() - start_time
            return result, depth, stats["nodes_explored"], elapsed, stats["max_stack_depth"]

        depth += 1

    elapsed = time.time() - start_time
    return None, None, stats["nodes_explored"], elapsed, stats["max_stack_depth"]

def main():
    print("=== 8-Puzzle Solver using Iterative Deepening DFS (IDDFS) ===")
    print("Name: Your Name")
    print("USN: Your USN")

    initial_state = get_matrix("STARTING")
    goal_state = get_matrix("GOAL")

    while True:
        try:
            max_limit = int(input("\nEnter a maximum depth limit for the search (e.g., 20 for a reasonable cap, 0 for no limit): "))
            if max_limit >= 0:
                break
            print("Please enter 0 or a positive integer.")
        except ValueError:
            print("Invalid input. Please enter an integer.")

    print("\n--- Initializing Search ---")
    print("Initial State:")
    print_state(initial_state)

    print("Goal State:")
    print_state(goal_state)

    limit_display = max_limit if max_limit > 0 else "No Limit"
    print(f"Max Depth Limit: {limit_display}")

    path, found_depth, total_nodes, exec_time, max_stack = solve_iddfs(
        initial_state, goal_state, max_limit
    )

    if path is not None:
        d = len(path) - 1
        b = 3
        print(f"\n✨ Success! Optimal solution found at depth: {found_depth}")
        print(f"Total moves required: {d}")
        print(f"Total nodes explored: {total_nodes}")

        print("\n--- Step-by-Step Path ---")
        for step, state in enumerate(path):
            if step == 0:
                print(f"Step {step} (Start):")
            else:
                print(f"Step {step}:")
            print_state(state)

        print("--- Empirical Performance (IDDFS) ---")
        print(f"Total Explored States (Nodes processed across all DLS iterations): {total_nodes}")
        print(f"Execution Time: {exec_time:.4f} seconds")
        print(f"Max Stack Depth for a DLS iteration (path length): {max_stack} nodes")

        print("\n--- Theoretical Complexities (IDDFS) ---")
        print(f"For Solution Depth (d) = {d}, Branching Factor (b) = {b}")
        print(f"Theoretical Time Complexity (Nodes explored, approximated O(b^d)): {b ** d} nodes (note: IDDFS explores nodes at shallower depths multiple times)")
        print(f"Theoretical Space Complexity (Max stack depth * branching factor, O(b*d)): {b ** d} nodes")

        print("\n--- General Complexity Analysis for Iterative Deepening DFS (IDDFS) ---")
        print("Time Complexity: O(b^d)")
        print("  'b' is the branching factor (average number of possible moves from a state). For the 8-puzzle, b is typically around 2-4.")
        print("  'd' is the depth of the shallowest goal state. IDDFS explores states up to depth 'd' multiple times.")
        print("Space Complexity: O(b^d)")
        print("  IDDFS performs a Depth-First Search at each iteration, so its space complexity is determined by the maximum depth of the DFS stack, which is 'd'.")
    else:
        print(f"\nNo solution found within maximum depth limit {max_limit}.")
        print(f"Total nodes explored: {total_nodes}")
        print(f"Execution Time: {exec_time:.4f} seconds")

if __name__ == "__main__":
    main()
```
