```python
def calculate_cost(state):
    """
    Computes the number of pairs of queens attacking each other
    directly or indirectly via row or diagonal alignments.
    """
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row check
            if state[i] == state[j]:
                attacks += 1
            # Diagonal check: if row difference equals column difference
            elif abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_neighbors_with_costs(state):
    """
    Generates all valid neighbor configurations by swapping two columns.
    Returns them ordered strictly by the lab's tie-breaking priority rule:
    Smallest column subscripts (i, j) are generated first.
    """
    neighbors = []
    n = len(state)

    # Nested loops ensure columns are paired in lexicographical order:
    # (0,1), (0,2), (0,3), (1,2), (1,3), (2,3)
    for i in range(n):
        for j in range(i + 1, n):
            neighbor_state = list(state)
            # Swap row positions of the two selected queens
            neighbor_state[i], neighbor_state[j] = neighbor_state[j], neighbor_state[i]
            cost = calculate_cost(neighbor_state)
            neighbors.append({
                'state': neighbor_state,
                'cost': cost,
                'swap': (i, j)
            })
    return neighbors

def hill_climbing_4_queens(initial_state):
    """
    Executes the hill-climbing search loop with explicit log printing
    to write directly into your lab observation notebook.
    """
    current_state = list(initial_state)
    current_cost = calculate_cost(current_state)

    step = 0
    print(f"Initial State State: {current_state} | Cost: {current_cost}\n")

    while True:
        print(f"--- Step {step}: Evaluating Neighbors ---")
        neighbors = get_neighbors_with_costs(current_state)

        best_neighbor = None

        # Display neighbor costs and look for improvements
        for nbr in neighbors:
            swap_info = f"Swap columns {nbr['swap']}"
            print(f"  Neighbor state: {nbr['state']} | Cost: {nbr['cost']} ({swap_info})")

            # We want to MINIMIZE cost. Find the first lowest cost neighbor.
            # Due to the way our neighbor loop runs, ties automatically favor the smallest index.
            if best_neighbor is None or nbr['cost'] < best_neighbor['cost']:
                best_neighbor = nbr

        print(f"\nBest found neighbor: {best_neighbor['state']} with Cost: {best_neighbor['cost']}")

        # In Hill-Climbing, if the best neighbor is NOT better than current, we stop.
        # Since we minimize conflicts, "not better" means neighbor cost >= current cost.
        if best_neighbor['cost'] >= current_cost:
            print("\n[Termination] No neighbor provides a better cost. Stopping algorithm.")
            break

        # Move to the better neighbor state
        current_state = best_neighbor['state']
        current_cost = best_neighbor['cost']
        step += 1
        print(f"Moved to state: {current_state} | New Cost: {current_cost}\n")

        # If cost is 0, we found a perfect solution!
        if current_cost == 0:
            print("--- Success! Perfect goal state reached. ---")
            break

    return current_state, current_cost

# --- EXECUTION WITH INITIAL VALUES FROM YOUR LAB SLIDE ---
if __name__ == "__main__":
    # Slide state given: x0=3, x1=1, x2=2, x3=0
    start_state = [3, 1, 2, 0]

    final_state, final_cost = hill_climbing_4_queens(start_state)

    print("\n==============================")
    print(f"Final State Result: {final_state}")
    print(f"Remaining Attacking Pairs: {final_cost}")
    print("==============================")
```
