```python
import time

def get_matrix(name):
    print(f"\nEnter the {name} (3x3 matrix, use 0 for blank tile, 3 numbers per line):")
    matrix = []
    for i in range(1, 4):
        while True:
            try:
                row = tuple(map(int, input(f"Row {i} (e.g., 1 2 3): ").strip().split()))
                if len(row) == 3:
                    matrix.append(row)
                    break
                else:
                    print("Please enter exactly 3 numbers separated by spaces.")
            except ValueError:
                print("Invalid input. Please enter integers only.")
    return tuple(matrix)

def print_state(state):
    for row in state:
        print(row)
    print("-" * 15)

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return -1, -1

def get_neighbors(state):
    r, c = find_blank(state)
    # Priority order: Up, Down, Left, Right
    moves = [
        ("Up", -1, 0),
        ("Down", 1, 0),
        ("Left", 0, -1),
        ("Right", 0, 1)
    ]
    neighbors = []
    for move_name, dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            grid = [list(row) for row in state]
            grid[r][c], grid[nr][nc] = grid[nr][nc], grid[r][c]
            new_state = tuple(tuple(row) for row in grid)
            neighbors.append((move_name, new_state))
    return neighbors

def solve_dfs(initial_state, goal_state, depth_limit=25):
    start_time = time.time()

    # Stack stores: (current_state, moves_list, states_list, current_depth)
    stack = [(initial_state, [], [initial_state], 0)]
    visited = {initial_state: 0}

    explored_states = 0
    max_memory = len(stack) + len(visited)

    while stack:
        max_memory = max(max_memory, len(stack) + len(visited))
        current_state, moves, path, depth = stack.pop()
        explored_states += 1

        if current_state == goal_state:
            elapsed_time = time.time() - start_time
            return moves, path, explored_states, elapsed_time, max_memory

        if depth < depth_limit:
            # Push in reverse order so 'Up' is popped first (LIFO)
            for move_name, next_state in reversed(get_neighbors(current_state)):
                next_depth = depth + 1
                if next_state not in visited or visited[next_state] > next_depth:
                    visited[next_state] = next_depth
                    stack.append((
                        next_state,
                        moves + [move_name],
                        path + [next_state],
                        next_depth
                    ))

    elapsed_time = time.time() - start_time
    return None, None, explored_states, elapsed_time, max_memory

def main():
    print("=== 8-PUZZLE DFS SOLVER ===")
    print("Name: Your Name")
    print("USN: Your USN")

    initial_state = get_matrix("INITIAL STATE")
    goal_state = get_matrix("GOAL STATE")

    print("\nInitial Configuration:")
    print_state(initial_state)

    print("Goal Configuration:")
    print_state(goal_state)

    depth_limit = 25
    print(f"Searching for a solution using DFS (Depth limit: {depth_limit})...\n")

    moves, path, explored, exec_time, max_mem = solve_dfs(initial_state, goal_state, depth_limit)

    if moves is not None:
        d = len(moves)
        b = 3  # Average branching factor for 8-puzzle
        print(f"Success! Goal reached in {d} steps.")
        print("Sequence of moves:", " -> ".join(moves))

        print("\nPath to Goal (showing states after each move):")
        print_state(path[0])
        for move, state in zip(moves, path[1:]):
            print(f"Move: {move}")
            print_state(state)

        print(f"\n--- Empirical Performance (DFS with depth limit {depth_limit}) ---")
        print(f"Total Explored States (Nodes processed by DFS): {explored}")
        print(f"Execution Time: {exec_time:.4f} seconds")
        print(f"Max Stack + Visited Set Size: {max_mem} nodes")

        print("\n--- Theoretical Complexities (O(b^d)) ---")
        print(f"For Solution Depth (d) = {d}, Branching Factor (b) = {b}")
        print(f"Theoretical Time Complexity (Nodes explored): {b ** d}")
        print(f"Theoretical Space Complexity (Nodes in memory): {b ** d}")
    else:
        print(f"No solution found within depth limit {depth_limit}.")
        print(f"Total Explored States: {explored}")
        print(f"Execution Time: {exec_time:.4f} seconds")

if __name__ == "__main__":
    main()
```
