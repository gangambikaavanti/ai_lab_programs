```python
import itertools

def evaluate_formula(expression, model):
    """Evaluates a logical expression string given a truth model dictionary."""
    # Replace tokens safely using a simple mapping
    eval_str = expression
    # Use explicit replacements for standard operators to match Python syntax
    eval_str = eval_str.replace("∨", " or ").replace("∧", " and ").replace("¬", " not ")

    # Replace variables with their Boolean string values
    for var, val in model.items():
        # Using spaces around variable handles simple standalone tokens cleanly
        eval_str = eval_str.replace(var, str(val))

    return bool(eval(eval_str))

def generate_truth_table():
    # Define the propositions and formulas from the image
    variables = ['A', 'B', 'C']
    f1 = "A ∨ C"
    f2 = "B ∨ ¬C"
    kb = f"({f1}) ∧ ({f2})"
    alpha = "A ∨ B"

    print(f"--- Truth Table Enumeration for KB ⊨ α ---")
    print(f"KB: {kb}")
    print(f"α: {alpha}\n")

    # Generate all combinations of True/False for A, B, C
    # To match the image order (False first), we use [False, True]
    combinations = list(itertools.product([False, True], repeat=3))

    # Header
    print(f"{'A':<7}{'B':<7}{'C':<7}{f1:<10}{f2:<10}{'KB':<8}{'α':<6}")
    print("-" * 55)

    entails = True

    for combo in combinations:
        model = dict(zip(variables, combo))

        # Evaluate each column intermediate step
        val_f1 = evaluate_formula(f1, model)
        val_f2 = evaluate_formula(f2, model)
        val_kb = evaluate_formula(kb, model)
        val_alpha = evaluate_formula(alpha, model)

        # Print row
        print(f"{str(model['A']):<7}{str(model['B']):<7}{str(model['C']):<7}"
              f"{str(val_f1):<10}{str(val_f2):<10}{str(val_kb):<8}{str(val_alpha):<6}")

        # Check entailment condition: If KB is True, alpha must be True
        if val_kb and not val_alpha:
            entails = False

    print("-" * 55)
    print(f"Conclusion: Does KB entail α? -> {entails}")

if __name__ == "__main__":
    generate_truth_table()
```
